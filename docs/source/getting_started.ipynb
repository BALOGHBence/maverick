## A minimal working example

Playing a game consists of

- Setting up players.
- Configuring a game.
- Playing the game.

There are ways to define custom players with their own behaviour and you will learn about how to do that in the User Guide. For now, we are going to define a Texas Hold'em game with a few simple bots from the library itself.

````python
# Import necessary modules from the maverick library
from maverick import Game, PlayerLike, PlayerState
from maverick.players import FoldBot, CallBot, AggressiveBot

# Create game with blinds and max hands
game = Game(small_blind=10, big_blind=20, max_hands=20)

# Create and add players with different strategies
players: list[PlayerLike] = [
    CallBot(name="CallBot"),
    AggressiveBot(name="AggroBot"),
    FoldBot(name="FoldBot"),
]

# Add players to the game
for player in players:
    game.add_player(player, state=PlayerState(stack=1000))

# Start the game
game.start()

# Inspect results — all players (including eliminated) are retained
print("All players:")
for snapshot in game.state.players:
    print(f"{snapshot.name} - Stack: {snapshot.state.stack} ({snapshot.state.state_type})")

# Active players only
print("\nActive players:")
for snapshot in game.state.get_active_players():
    print(f"{snapshot.name} is still in the game with {snapshot.state.stack} chips")

# Non-eliminated players
print("\nNon-eliminated players:")
for snapshot in game.state.get_non_eliminated_players():
    print(f"{snapshot.name} is eligible for the next hand with {snapshot.state.stack} chips")

# Announce winner
print("\nAnd the winner is...")
winner = max(game.state.players, key=lambda p: p.state.stack)
print(f"The winner is {winner.name} with a stack of {winner.state.stack}!")
````
